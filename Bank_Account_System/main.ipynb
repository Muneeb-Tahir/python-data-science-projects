Challenge 6 — Bank Account System

Create classes:

BankAccount
SavingsAccount
CurrentAccount
Customer

Implement:

deposit()
withdraw()
check_balance()
transfer()

Rules:

Cannot withdraw more than balance
Savings account has interest
Current account can have an overdraft
Every transaction should be recorded

Example:

account.deposit(50000)
account.withdraw(5000)
account.transfer(other_account, 10000)

Then print:

Transaction History

Deposit       +50,000
Withdrawal     -5,000
Transfer      -10,000
----------------------
Balance        35,000

This will force you to actually understand OOP instead of just knowing the syntax.

```python
class bank_account:

    def __init__(self, account_number, account_holder, balance=0):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance
        self.transaction_history = []

    def deposit(self, amount):

        if amount > 0:
            self.balance += amount

            self.transaction_history.append(("Deposit", amount))

            print(f"Deposited ${amount}. New balance: ${self.balance}.")

        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):

        if 0 < amount <= self.balance:
            self.balance -= amount

            self.transaction_history.append(("Withdrawal", -amount))

            print(f"Withdrew ${amount}. New balance: ${self.balance}.")

        else:
            print("Withdrawal amount must be positive "
                  "and less than or equal to the current balance.")

    def check_balance(self):
        return self.balance

    def transfer(self, amount, target_account):

        if 0 < amount <= self.balance:

            self.balance -= amount

            target_account.deposit(amount)

            self.transaction_history.append(("Transfer", -amount))

            print(f"Transferred ${amount} to "
                  f"{target_account.account_holder}. "
                  f"New balance: ${self.balance}.")

        else:
            print("Transfer amount must be positive "
                  "and less than or equal to the current balance.")

    def show_transaction_history(self):

        print("\nTransaction History")

        for transaction, amount in self.transaction_history:
            print(f"{transaction:<15} {amount:+,}")

        print("-" * 30)
        print(f"{'Balance':<15} {self.balance:,}")


bank_account1 = bank_account("123456", "Alice", 1000)
bank_account2 = bank_account("654321", "Bob", 500)
bank_account1.deposit(200)
bank_account1.withdraw(150)
bank_account1.transfer(100, bank_account2)
bank_account1.show_transaction_history()
bank_account2.show_transaction_history()
```

```python
class savings_account(bank_account):

    def __init__(
        self,
        account_number,
        account_holder,
        balance=0,
        interest_rate=0.01
    ):
        super().__init__(
            account_number,
            account_holder,
            balance
        )

        self.interest_rate = interest_rate

    def add_interest(self):

        interest = self.balance * self.interest_rate

        self.balance += interest

        self.transaction_history.append(
            ("Interest", interest)
        )

        print(
            f"Added interest: ${interest}. "
            f"New balance: ${self.balance}."
        )


savings_account1 = savings_account("789012", "Charlie", 2000, 0.02)
savings_account1.add_interest()
savings_account1.show_transaction_history()
savings_account1.deposit(500)
savings_account1.withdraw(300)
savings_account1.show_transaction_history()
```

```python
class current_account(bank_account):

    def __init__(self,
                 account_number,
                 account_holder,
                 balance=0,
                 overdraft_limit=500):
        super().__init__(account_number, account_holder, balance)

        self.overdraft_limit = overdraft_limit

    def withdraw(self, amount):

        if 0 < amount <= self.balance + self.overdraft_limit:

            self.balance -= amount

            self.transaction_history.append(("Withdrawal", -amount))

            print(f"Withdrew ${amount}. "
                  f"New balance: ${self.balance}.")

        else:
            print("Withdrawal amount exceeds the overdraft limit.")


current_account1 = current_account("345678", "David", 300, 200)
current_account1.withdraw(400)
current_account1.withdraw(200)
current_account1.deposit(100)
current_account1.show_transaction_history()
```

```python
class customer:

    def __init__(self, name):
        self.name = name
        self.accounts = []

    def add_account(self, account):

        self.accounts.append(account)

        print(f"Account {account.account_number} "
              f"added for {self.name}.")

    def get_total_balance(self):

        total_balance = sum(account.check_balance()
                            for account in self.accounts)

        return total_balance

    def show_transaction_history(self):

        print(f"\nTransaction history for {self.name}:")

        for account in self.accounts:

            print(f"\nAccount {account.account_number}:")

            for transaction, amount in account.transaction_history:
                print(f"{transaction:<15} {amount:+,}")
                print(f"{account.check_balance():<15} {amount:+,}")
customer1 = customer("John Doe")
savings_acc = savings_account("SA123", "John Doe", 1000, 0.02)
current_acc = current_account("CA123", "John Doe", 500, 1000)
customer1.add_account(savings_acc)
customer1.add_account(current_acc)
customer1.accounts[0].deposit(200)
customer1.accounts[0].withdraw(100)
customer1.accounts[1].withdraw(200)
customer1.accounts[0].add_interest()
customer1.accounts[0].transfer(150, customer1.accounts[1])
# customer1.accounts[1].withdraw(800)  # This should be allowed due to overdraft
# customer1.accounts[1].withdraw(300)  # This should exceed the overdraft limit
customer1.accounts[0].show_transaction_history()
customer1.accounts[1].show_transaction_history()
# customer1.show_transaction_history()
```
