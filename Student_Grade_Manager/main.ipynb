## Challenge 1 — Student Grade Manager

Create a program that stores students and their marks.

Example:

students = {
    "Ali": [78, 85, 91, 67],
    "Ahmed": [90, 88, 76, 95],
    "Sara": [92, 91, 89, 94]
}

Your program should calculate:

Each student's average
Highest mark ,
Lowest mark ,
Grade ,
Class average ,
Highest-performing student ,
Number of students who passed ,
Number of students who failed ,

Bonus: Don't use NumPy or Pandas.

```python
students = {
    "Ali": [78, 85, 91, 67],
    "Ahmed": [90, 88, 76, 95],
    "Sara": [92, 91, 89, 94]
}
print("Student Grades:", students)
for name, grades in students.items():
    avg = sum(grades) / len(grades)
    highest_marks = max(grades)
    lowest_marks = min(grades)
    class_average = sum(sum(grades) for grades in students.values()) / sum(len(grades) for grades in students.values())
    Highest_performing_student = max(students, key=lambda x: sum(students[x]) / len(students[x]))
    No_of_students_who_passed = sum(1 for grades in students.values() if sum(grades) / len(grades) >= 60)
    No_of_students_who_failed = sum(1 for grades in students.values() if sum(grades) / len(grades) < 60)


    print("Student Name: {}, Average Grade: {:.2f}, Highest Marks: {}, Lowest Marks: {}, Class Average: {:.2f}, Highest Performing Student: {}, Number of Students Who Passed: {}, Number of Students Who Failed: {}".format(
        name, avg, highest_marks, lowest_marks, class_average, Highest_performing_student, No_of_students_who_passed, No_of_students_who_failed))
    #{:.2f} is used to format the average grade to two decimal places.
```
