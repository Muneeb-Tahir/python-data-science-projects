Challenge 4 — Contact Management System

Build:

1. Add contact
2. Search contact
3. Update contact
4. Delete contact
5. Display all contacts
6. Exit

Store information like:

{
    "Ali": {
        "phone": "03001234567",
        "email": "ali@gmail.com"
    }
}

Add validation:

Phone number must be valid
Email must contain @
Don't allow duplicate names

This is a good exercise before moving into OOP.

```python
"""
Challenge 4 — Contact Management System

Build:

1. Add contact
2. Search contact
3. Update contact
4. Delete contact
5. Display all contacts
6. Exit


Add validation:

Phone number must be valid
Email must contain @
Don't allow duplicate names

"""

Information = {}
while True:
    print("Add Contact: ")
    print("Search Contact: ")
    print("Update Contact: ")
    print("Delete Contact: ")
    print("Display All Contacts: ")
    print("Exit: ")
    choice = int(input("Enter your choice: "))
    if choice == 1:
        # Add contact logic here
        add=input("Enter contact name: ")
        phone = input("Enter phone number: ")
        email = input("Enter email: ")
        if email and "@" not in email:
            print("Invalid email format. Please include '@' in the email.")
            continue
        if phone and not phone.isdigit():
            print("Invalid phone number. Please enter digits only.")
            continue
        no_duplicate = True
        for name in Information:
            if name == add:
                no_duplicate = False
                break
        if not no_duplicate:
            print("Contact with this name already exists.")
            continue
        if phone and email:
            Information[add] = {"phone": phone, "email": email}
            print(f"Contact {add} added.")
        else:
            print("Please provide both phone number and email.")

    elif choice == 2:
        # Search contact logic here
        search_name = input("Enter contact name to search: ")
        if search_name in Information:
            contact_info = Information[search_name]
            print(f"Name: {search_name}, Phone: {contact_info['phone']}, Email: {contact_info['email']}")
        else:
            print(f"Contact {search_name} not found.")
    elif choice == 3:
        # Update contact logic here
        update_name = input("Enter contact name to update: ")
        if update_name in Information:
            new_phone = input("Enter new phone number: ")
            new_email = input("Enter new email: ")
            if new_email and "@" not in new_email:
                print("Invalid email format. Please include '@' in the email.")
                continue
            if new_phone and not new_phone.isdigit():
                print("Invalid phone number. Please enter digits only.")
                continue
            Information[update_name] = {"phone": new_phone, "email": new_email}
            print(f"Contact {update_name} updated.")
        else:
            print(f"Contact {update_name} not found.")

    elif choice == 4:
        # Delete contact logic here
        delete_name = input("Enter contact name to delete: ")
        if delete_name in Information:
            del Information[delete_name]
            print(f"Contact {delete_name} deleted.")
        else:
            print(f"Contact {delete_name} not found.")
    elif choice == 5:
        # Display all contacts logic here
        if Information:
            for name, info in Information.items():
                print(f"Name: {name}, Phone: {info['phone']}, Email: {info['email']}")
        else:
            print("No contacts to display.")
    elif choice == 6:
        print("Exiting the program.")
        break
    else:
        print("Invalid choice. Please try again.")

```
