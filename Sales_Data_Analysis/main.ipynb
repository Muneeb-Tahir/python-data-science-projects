```python
import pandas as pd
import numpy as np


df=pd.read_csv("D:/Code PlayGround/AI_ML_Prac/Sales_Data_Analysis/sales_data_100_rows.csv")
print(df.head())
print("all rows and columns in the dataset:", df.shape)
# pd.set_option("display.max_rows", None)

# print(df)
```

```python
total_revenue = df['Price'].sum()
print(f"Total Revenue: {total_revenue}")
```

```python
avg_order_value = df['Price'].mean()
print(f"Average Order Value: {avg_order_value}")
```

```python
most_sold_product = df['Product'].mode()[0]
print(f"Most Sold Product: {most_sold_product}")
```

```python
category_generated_most_revenue = df.groupby('Category')['Price'].sum().idxmax()
each_category_revenue = df.groupby('Category')['Price'].sum()
print(f"Revenue generated by each category:\n{each_category_revenue}")
print(f"Category that generated the most revenue: {category_generated_most_revenue}")
```

```python
category_generated_most_revenue = df.groupby('City')['Price'].sum().idxmax()
each_category_revenue = df.groupby('City')['Price'].sum()
print(f"Revenue generated by each City:\n{each_category_revenue}")
print(f"City that generated the most revenue: {category_generated_most_revenue}")
```

```python
monthly_revenue = df.groupby('Date')['Price'].sum()
print(f"Revenue generated by each Date:\n{monthly_revenue}")
Overall_revenue_Date = df.groupby('Date')['Price'].sum().idxmax()
print(f"Date that generated the most revenue: {Overall_revenue_Date}")
```

```python
top_10_customers = df.groupby('Customer')['Price'].sum().nlargest(10)
print(f"Top 10 Customers by Revenue:\n{top_10_customers}")
```

```python
avg_quantity_per_order = df['Quantity'].mean()
print(f"Average Quantity per Order: {avg_quantity_per_order}")
```

```python
declining_products = df.groupby('Product')['Quantity'].sum().nsmallest(10)
print(f"Products with declining sales:\n{declining_products}")
declining_products = df.groupby('Product')['Price'].sum().nsmallest(10)
print(f"Products with declining revenue:\n{declining_products}")
```

```python
percentage_revenue_per_category = (df.groupby('Category')['Price'].sum() / total_revenue) * 100
print(f"Percentage of Revenue per Category:\n{percentage_revenue_per_category}")
```
