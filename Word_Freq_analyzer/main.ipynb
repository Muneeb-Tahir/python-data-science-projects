Challenge 3 — Word Frequency Analyzer

Give your program a paragraph:

Python is easy to learn. Python is powerful.
Python is used for data science and machine learning.

Your program should output:

python: 3
is: 3
to: 1
learn: 1
powerful: 1
...

Then implement:

Remove punctuation
Convert everything to lowercase
Ignore common words
Find the most common word
Find the 10 most common words



```python
String = (
    "Python is easy to learn. Python is powerful. Python is used for data science and machine learning."
)

#Remove punctuation
String = String.replace(".", "").replace(",", "").replace("!", "").replace("?", "")

print("After removing punctuation: ", String)
#Convert everything to lowercase
String = String.lower()
print("After converting to lowercase: ", String)
#Ignore common words
common_words = {"is", "for", "and", "the", "a", "an"}
print("Common words to ignore: ", common_words)
#Find the most common word

most_common_word = ""
highest_count = 0

for word in String.split():
    if word not in common_words:
        count = String.count(word)
        if count > highest_count:
            highest_count = count
            most_common_word = word

print("Most common:", most_common_word, highest_count)
#Find the 10 most common words

word_counts = {}
for word in String.split():
    if word not in common_words:
        count = String.split().count(word)
        word_counts[word] = count

most_common_10 = sorted(
    word_counts.items(),
    key=lambda x: x[1],
    reverse=True
)[:10]

print("10 most common words:")

for word, count in most_common_10:
    print(word, count)


```
